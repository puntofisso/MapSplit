```python
# Local authority files
# LA-coords.csv
# LA-population.csv

import csv
import pandas as pd
import numpy as np

df_coords = pd.read_csv("LA-coords.csv")
df_population = pd.read_csv("LA-population.csv")
```

```python
df_coords
```

```python
df_population
```

```python
df = pd.merge(df_coords, df_population, left_on='Code', right_on='Code', how='left')
df = df.drop(columns=['Name_y'])
df = df.rename(columns={"Name_x": "Name"})
```

```python
df
```

```python
df.to_csv("LA.csv")
```

```python
# LSOA files (England + Wales)
# LSOA-coords.csv
# LSOA-population.csv
import csv
import pandas as pd
import numpy as np

df_lsoa_coords = pd.read_csv("LSOA-coords.csv")
df_lsoa_population = pd.read_csv("LSOA-population.csv")
```

```python
df_lsoa_coords
```

```python
df_lsoa_population
```

```python
df_lsoa = pd.merge(df_lsoa_coords, df_lsoa_population, left_on='LSOA21CD', right_on='LSOA 2021 Code', how='left')
df_lsoa = df_lsoa.drop(columns=['LSOA 2021 Name','LSOA 2021 Code','FID','LSOA21NMW','BNG_E','BNG_N','Shape__Area','Shape__Length','GlobalID'])
df_lsoa = df_lsoa.rename(columns={"Total": "Population"})
```

```python
df_lsoa.to_csv('LSOA.csv')
```

```python
# Local authority income files
# LA-coords.csv
# LA-income.csv

import csv
import pandas as pd
import numpy as np

df_coords = pd.read_csv("LA-coords.csv")
df_income = pd.read_csv("LA-income.csv")
```

```python
df_coords
```

```python
df_income
```

```python
df_income = pd.merge(df_coords, df_income, left_on='Code', right_on='Area Code', how='left')
df_income = df_income.drop(columns=['Area Code','Area Name'])
```

```python
df_income
```

```python
df_income.to_csv('HMRC.csv')
```
